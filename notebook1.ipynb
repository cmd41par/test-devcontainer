```python
import pandas as pd
import matplotlib.pyplot as plt

# Create a small DataFrame
data = {
    'Category': ['A', 'B', 'C', 'D'],
    'Values': [10, 20, 15, 25]
}
df = pd.DataFrame(data)

# Plot the DataFrame
plt.figure(figsize=(8, 6))
plt.bar(df['Category'], df['Values'], color='skyblue')
plt.xlabel('Category')
plt.ylabel('Values')
plt.title('Category vs Values')
plt.show()
```
